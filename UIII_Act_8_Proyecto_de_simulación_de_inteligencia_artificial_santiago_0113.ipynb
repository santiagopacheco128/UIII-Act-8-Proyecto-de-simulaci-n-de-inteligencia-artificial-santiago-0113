```python
print("🥖 Sistema Inteligente de Panadero y Horno 🥖")

while True:

    print("\n====== MENÚ PRINCIPAL ======")
    print("1. Encender horno")
    print("2. Hornear pan")
    print("3. Revisar temperatura")
    print("4. Preparar masa")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")

    # OPCIÓN 1
    if opcion == "1":
        temperatura = int(input("Ingresa la temperatura del horno: "))

        if temperatura >= 180:
            print("🔥 El horno está listo para hornear.")
        elif temperatura >= 100:
            print("⚠️ El horno aún se está calentando.")
        else:
            print("❄️ El horno está muy frío.")

    # OPCIÓN 2
    elif opcion == "2":
        tipo = input("¿Qué tipo de pan deseas hornear? ")

        if tipo == "bolillo":
            print("🥖 Horneando bolillos...")
        elif tipo == "pastel":
            print("🎂 Horneando pastel...")
        elif tipo == "pizza":
            print("🍕 Horneando pizza...")
        else:
            print("❌ Tipo de pan no disponible.")

    # OPCIÓN 3
    elif opcion == "3":
        temperatura = int(input("Ingresa la temperatura actual: "))

        if temperatura > 200:
            print("🔥 Mucho calor, baja la temperatura.")
        elif temperatura >= 150:
            print("✅ Temperatura perfecta.")
        else:
            print("⚠️ Temperatura baja.")

    # OPCIÓN 4
    elif opcion == "4":
        harina = input("¿Hay harina disponible? (si/no): ")

        if harina == "si":
            print("🥣 Preparando masa...")
        else:
            print("❌ No hay harina suficiente.")

    # OPCIÓN 5
    elif opcion == "5":
        print("👋 Saliendo del sistema del panadero...")
        break

    # OPCIÓN INVÁLIDA
    else:
        print("❌ Opción no válida.")
```
